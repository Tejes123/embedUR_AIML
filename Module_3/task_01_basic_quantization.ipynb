# INT8 affine quantization

```python
import numpy as np
```

```python
tensor = np.array(
    [1.0, -0.5, 3.2, -2.8, 0.0],
    dtype=np.float32
)

scale = 0.025
zero_point = 0
```

```python
def quantize_tensor(tensor, scale, zero_point):
    """
    Affine INT8 quantization

    q = round(x / scale) + zero_point
    clip to [-128, 127]
    """

    q = np.round(tensor / scale) + zero_point

    q = np.clip(q, -128, 127)

    return q.astype(np.int8)

def dequantize_tensor(quantized_tensor, scale, zero_point):
    """
    Affine dequantization

    x = (q - zero_point) * scale
    """

    return (quantized_tensor.astype(np.float32) - zero_point) * scale
```

## Run Quantization and Dequantization

```python
quantized_tensor = quantize_tensor(
    tensor,
    scale,
    zero_point
)

dequantized_tensor = dequantize_tensor(
    quantized_tensor,
    scale,
    zero_point
)
```

## Compute Error Metrics

```python
absolute_error = np.abs(
    tensor - dequantized_tensor
)

mae = np.mean(absolute_error)

max_error = np.max(absolute_error)
```

```python
print("Original tensor:")
print(tensor)

print("\nQuantized tensor (INT8):")
print(quantized_tensor)

print("\nDequantized tensor (float):")
print(dequantized_tensor)

print("\nAbsolute error per element:")
print(absolute_error)

print("\nMean Absolute Error (MAE):")
print(mae)

print("\nMaximum Absolute Error:")
print(max_error)
```
