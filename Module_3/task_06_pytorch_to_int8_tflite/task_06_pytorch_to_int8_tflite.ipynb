## Task 6: Convert a PyTorch .pth Model to Fully INT8 TFLite

Run once after training.

```python
import os
import numpy as np

from torchvision import datasets
from torchvision import transforms

transform = transforms.Compose([
    transforms.ToTensor()
])

test_data = datasets.MNIST(
    "./data",
    train=False,
    download=True,
    transform=transform
)

os.makedirs(
    "calib",
    exist_ok=True
)

for i in range(100):

    sample, _ = test_data[i]

    np.save(
        f"calib/{i}.npy",
        sample.numpy()
    )

print("Calibration samples saved")
```
